# Reproduce all analyses

This notebook runs the version-controlled scripts used for participant-wise correlations, Fisher z inference, GEE models, cross-classified mixed models, manuscript figures, the non-linearity (uncanny-valley) test, and the automated facial-expression robustness analysis.

## 1. Locate the repository and check the environment

```python
from pathlib import Path
import subprocess
import sys

REPO_ROOT = Path.cwd().resolve()
if not (REPO_ROOT / 'src').exists():
    REPO_ROOT = REPO_ROOT.parent
if not (REPO_ROOT / 'src').exists():
    raise RuntimeError('Start Jupyter from the repository root or notebooks directory.')
print('Repository:', REPO_ROOT)
print('Python:', sys.version)
```

## 2. Validate the public analysis data and stimulus files

```python
subprocess.run([sys.executable, str(REPO_ROOT / 'src' / 'validate_data.py')], check=True)
subprocess.run([sys.executable, str(REPO_ROOT / 'src' / 'validate_stimuli.py')], check=True)
```

## 3. Participant-wise Spearman correlations and Fisher's z inference

```python
subprocess.run([sys.executable, str(REPO_ROOT / 'src' / 'correlations.py')], check=True)
```

## 4. GEE analyses and marginal predictions

```python
subprocess.run([sys.executable, str(REPO_ROOT / 'src' / 'gee_analysis.py')], check=True)
```

## 5. Cross-classified mixed models

This step uses all 50,000 observations and can take several minutes.

```python
subprocess.run([sys.executable, str(REPO_ROOT / 'src' / 'mixed_models.py')], check=True)
```

## 6. Manuscript figures without face images

```python
subprocess.run([sys.executable, str(REPO_ROOT / 'src' / 'figures.py')], check=True)
```

## 7. Non-linearity (uncanny-valley) test

Tests whether perceived artificiality relates to trustworthiness and attractiveness non-linearly, as an uncanny-valley dip would predict, or monotonically. Uses only the public ratings and the base environment.

```python
subprocess.run([sys.executable, str(REPO_ROOT / 'src' / 'nonlinearity_analysis.py')], check=True)
```

## 8. Automated facial-expression robustness analysis

The included image-derived features are sufficient to reproduce the adjustment. This step fits the GEE sensitivity models and the AU12-adjusted cross-classified models; it does not rerun Py-Feat image extraction.

```python
subprocess.run([sys.executable, str(REPO_ROOT / 'src' / 'expression_analysis.py')], check=True)
```
